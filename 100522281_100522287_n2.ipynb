**5.4 Predecir las instancias de bank_competition y guardarlo en CSV**

Para predecir las salidas de bank_competition y guardarlo en el formato csv, ejecutamos la siguiente celda:

```python

import pandas as pd
from sklearn import svm
import pickle

# Cargamos el modelo final que hicimos en el Cuaderno 1
nombre_modelo = 'modelo_final.pkl'
with open(nombre_modelo, 'rb') as archivo:
    modelo_cargado = pickle.load(archivo)

print(f"Modelo '{nombre_modelo}' cargado correctamente en memoria.")

# Cargamos los datos crudos de la competición
archivo_competicion = 'bank_competition.pkl'
df_competicion = pd.read_pickle(archivo_competicion)

print(f"Datos de competición cargados. Total de clientes a predecir: {len(df_competicion)}")

# Generamos las predicciones
predicciones = modelo_cargado.predict(df_competicion)

# 4. Guardamos las predicciones en el CSV solicitado
df_predicciones = pd.DataFrame(predicciones, columns=['prediction'])
df_predicciones.to_csv('predicciones.csv', index=False)

print("Archivo 'predicciones.csv' generado con éxito.")
```
